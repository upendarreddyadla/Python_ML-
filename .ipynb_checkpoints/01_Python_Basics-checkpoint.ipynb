## Control flow

### if / elif / else

```python
if, elif, and else are conditional statements in Python used to make decisions based on whether a condition is True or False

if → Checks the first condition
elif → Checks another condition if the previous condition is False
else → Executes when all the above conditions are False
```

```python
age = 20                                 #age >= 18 → True, so it prints Adult
                                          Python then stops and does not check elif or else
if age >= 18:
    print("Adult")

elif age >= 13:
    print("Teenager")

else:
    print("Child")
```

### For Loop

```python
A for loop in Python is used when you want to repeat a block of code a fixed number of times or go through each item in a collection 
like a list, string, or range of numbers
```

```python
for i in range(5):
    print(i)
```

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

```python
name = "Ajay"

for letter in name:
    print(letter)
```

### While Loop

```python
A while loop in Python is used when you want to repeat a block of code as long as a condition is true

while → checks a condition before each loop
If the condition is True, the code inside runs
If the condition becomes False, the loop stops
```

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

### Break

```python
The break statement in Python is used inside loops to stop the loop immediately. As soon as Python sees break,
it jumps out of the loop and continues with the code after the loop
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

### Continue

```python
The continue statement is used inside a loop to skip the current iteration and move to the next iteration
```

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

### Pass

```python
The pass statement is used when you want to leave a block of code empty without causing an error
```

```python
for i in range(5):
    if i == 2:
        pass
    print(i)
```

## Functions

```python
A function is a reusable block of code that performs a particular task

Suppose you need to calculate the sum many times. Instead of writing the same code repeatedly,
you can create a function once and use it whenever needed

To create a function, we use the def keyword
def function_name():
    # code
```

```python
def add():            # def is a keyword,add is a function name,add() calling the function
    a = 10
    b = 20
    print(a + b)

add()
```

```python
def add(a, b):                        #we can pass more than two Arguments
    print(a + b)

add(10, 20)
```

```python
A default argument is a parameter that already has a value

If the user doesn't provide a value, Python uses the default value
```

```python
def greet(name="Guest"):           #we didn't provide any name python uses guest 
    print("Hello", name)

greet()
```

```python
def greet(name="Guest"):
    print("Hello", name)

greet("Ravi")
```

### Variable-Length Arguments (*args, **kwargs)

```python
Sometimes we don't know how many arguments a function will receive
For example, you might want a function that can accept

2 numbers
5 numbers
10 numbers
100 numbers

Python provides:

*args
**kwargs

*args allows a function to accept any number of positional arguments
**kwargs allows a function to accept any number of keyword arguments
```

```python
def add(*args):
    total = 0

    for number in args:
        total = total + number

    print(total)

add(10, 20, 30, 40)
```

```python
def details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

details(
    name="Ravi",
    age=25,
    city="Hyderabad"
)
```

### Lambda Functions

```python
A lambda function is a small anonymous function written in a single expression

Lambda functions are useful when you need a small function for a short operation, especially with functions such as map(), filter(), and sorted()

Syntax:lambda arguments: expression
```

```python
square = lambda x: x * x              #Lambda is a keyword #X is argument,X*2 is a Expression

print(square(5))

add = lambda a, b: a + b

print(add(10, 20))
```

### Recurcive Functions

```python
A recursive function is a function that calls itself
```

```python
def countdown(n):
    if n == 0:
        print("Done")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```

```python
def factorial(n):
    if n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```
