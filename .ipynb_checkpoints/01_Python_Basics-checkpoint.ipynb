### Variables

### Datatypes

```python
A Datatype specifies what type of value srored in a variable 
like numbers,Text etc

Datatypes are

int-it stores whole numbers
float-it stores decimal numbers
str-it stores Text
bool-True or False
```

```python
Age = 25  #Age is a Variable name 25 is a integer
print(Age)

num = 12.5 #num is a Variable name 12.5 is a float value
print(num)

Name = "UpendarReddy" #Name is variable name UpendarReddy is a str value
print(Name)

is_variable = True  #is_variable is variable name True is Boolean Value
print(is_variable)
```
