### Datatypes

```python
A Datatype specifies what type of value srored in a variable 
like numbers,Text etc

Datatypes are

int-it stores whole numbers
float-it stores decimal numbers
str-it stores Text
bool-True or False
```

```python
Age = 25  #Age is a Variable name 25 is a integer
print(Age)

num = 12.5 #num is a Variable name 12.5 is a float value
print(num)

Name = "UpendarReddy" #Name is variable name UpendarReddy is a str value
print(Name)

is_variable = True  #is_variable is variable name True is Boolean Value
print(is_variable)
```

### Operators

```python
An operator is a special symbol used to perform an operation on values or variables

Types of Operators:
    
Arithmetic Operators → +, -, *, /, %, //, **
Comparison Operators → ==, !=, >, <, >=, <=
Assignment Operators → =, +=, -=, *=, /=
```

```python
a = 4              # Arithmetic operators
b = 2

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a % b)   # Modulus
print(a // b)  # Floor Division
print(a ** b)  # Power
```

```python
a = 10         #Comparison operators
b = 5

print(a == b)      
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
x = 10            #Assignment Operators

x += 5
print(x)    

x -= 3
print(x)    

x *= 2
print(x)    

x /= 4
print(x)    
```
