# Notebook 4 – Logging

### What is Logging?


```python
Logging in python is the process of recording events that happens during the Execution ,Such as errors,Warnings or status updates,
using the built-in logging module.
it is more powerful and flexible than print() because you can set severity levels,format messages and save logs to file for later Analysis

Tracks program flow: Helps developers understand what the program is doing

Debugging: Easier to find where and why errors occur

Monitoring: Useful in production systems to monitor performance and issues

Persistence: Logs can be saved to files for future analysis
```

### why use Logging?

```python
1.Better than print()

print() only shows output on the console

Logging can record events with different severity levels (DEBUG, INFO, WARNING, ERROR, CRITICAL)


2.Persistent Records

Logs can be saved to a file for later analysis

Useful for debugging issues that happen in production systems


3.Structured Information

Logging messages can include timestamps, severity, module name, etc

Makes it easier to trace what happened and when


4.Configurable

You can control what gets logged (e.g., only WARNING and above)

You can log to multiple places at once (console, file, remote server)


5.Professional Practice

In real projects, logging is a best practice

It helps developers, testers, and system admins monitor applications
```

### Log levels

```python
1.DEBUG: Used when you need detailed information while developing or debugging
```

```python
import logging

logging.basicConfig(level=logging.DEBUG)

def add_numbers(a, b):
    logging.debug(f"Adding numbers: a={a}, b={b}")  
    return a + b

result = add_numbers(5, 7)
print("Result:", result)
```

```python
2.INFO: Used for normal program information
```

```python
logging.info("User logged in successfully") #Example
```

```python
3.WARNING: Used when something might be a problem, but the program can continue
```

```python
age = 17           #Example

if age < 18:
    logging.warning("User is under 18")
```

```python
4.ERROR: Used when an error occurs
```

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Cannot divide by zero")
```

```python
5.CRITICAL: Used for a very serious problem that may stop the application
```

```python
logging.critical("Database connection failed") #Example
```

```python
Think of it like this:

DEBUG → INFO → WARNING → ERROR → CRITICAL

DEBUG → What is happening internally?
INFO → What is happening normally?
WARNING → Something unusual happened
ERROR → Something failed
CRITICAL → Very serious failure
```

```python
#Example

logging.debug("Checking account details")
logging.info("User logged in")
logging.warning("Balance is low")
logging.error("Payment failed")
logging.critical("Bank database is down")
```

### Log Formatters


```python
A formatter defines the layout/structure of log messages

It controls what information appears (like time, level, message, filename, line number)

Without formatters, logs are plain and less useful
```

```python
#Example
import logging

logger = logging.getLogger("my_logger")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)


logger.addHandler(console_handler)

logger.debug("This is a debug message")
logger.info("Application started successfully")
logger.warning("Low disk space warning")
logger.error("An error occurred")
logger.critical("Critical failure!")
```

```python
Common Formatter Fields:-
    
%(asctime)s → Timestamp

%(name)s → Logger name

%(levelname)s → Severity level

%(message)s → Log message

%(filename)s → File name

%(lineno)d → Line number
```

### File Handlers


```python
A FileHandler is a logging component that writes log messages to a file instead of just showing them on the console

It’s part of the logging module

Useful when you want to keep a permanent record of logs for debugging, monitoring, or auditing
```

```python
#Example

import logging

logger = logging.getLogger("my_logger")
logger.setLevel(logging.DEBUG)

file_handler = logging.FileHandler("app.log")
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")


file_handler.setFormatter(formatter)


logger.addHandler(file_handler)


logger.debug("Debugging details here")
logger.info("Application started successfully")
logger.warning("Low disk space warning")
logger.error("An error occurred")
logger.critical("Critical failure!")
```

```python
Why Use File Handlers?

Persistent logs → Saved to disk for later analysis

Separation of concerns → Console shows quick info, file keeps detailed records

Production monitoring → Helps track issues even after program ends

Flexibility → You can use multiple handlers (e.g., one for console, one for file)
```

### Console Handlers

```python
A ConsoleHandler (StreamHandler) sends log messages to the console (stdout)

By default, logging.basicConfig() sets up a console handler for you

But you can create and customize your own handler to control formatting and log levels
```

```python
#Example
import logging

logger = logging.getLogger("my_logger")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()

formatter = logging.Formatter("%(levelname)s - %(message)s")
console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.debug("This is a debug message")
logger.info("Application started successfully")
logger.warning("Low disk space warning")
logger.error("An error occurred")
logger.critical("Critical failure!")
```

```python
Why Use Console Handlers?

Quick feedback → See logs immediately while running the program

Custom formatting → Control how logs look in the console

Combine with FileHandler → Console shows live updates, file keeps permanent records
```

### Rotating Logs (Bonus)

```python
In long‑running applications, log files can grow very large

Rotating logs automatically create new log files when the current one reaches a certain size or time interval

This prevents logs from becoming too big and keeps them organized

Types of Rotating Handlers:
    
RotatingFileHandler → rotates logs based on file size

TimedRotatingFileHandler → rotates logs based on time intervals (daily, hourly, etc.)
```

```python
#Example
import logging
from logging.handlers import RotatingFileHandler


logger = logging.getLogger("size_logger")
logger.setLevel(logging.DEBUG)

handler = RotatingFileHandler("size_app.log", maxBytes=1024, backupCount=3)

formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(100):
    logger.debug(f"Log entry {i}")
```

```python
Why Use Rotating Logs?

Prevents log files from growing too large

Keeps logs organized and manageable

Useful in production systems where logs run continuously

Makes debugging easier because logs are split into chunks
```

### Logging Best Practices

```python
1. Use Appropriate Log Levels

DEBUG → Detailed developer info

INFO → General application events

WARNING → Something unexpected, but not fatal

ERROR → A serious problem occurred

CRITICAL → Severe issue, program may not continue

2.Always Use Formatters

Include timestamps, severity, and messages

Ex format: "%(asctime)s - %(levelname)s - %(message)s"

3.Separate Console and File Logs

Use ConsoleHandler for quick feedback

Use FileHandler (or RotatingFileHandler) for permanent records

Ex: Console shows WARNING+, file stores everything (DEBUG+)

4.Avoid print() for Debugging

print() disappears after program ends

Logging is persistent, configurable, and professional

5.Rotate Logs

Use RotatingFileHandler or TimedRotatingFileHandler to prevent huge files

Keeps logs manageable and organized

6.Don’t Log Sensitive Data

Avoid logging passwords, API keys, or personal information

Logs may be shared or stored long‑term

7.Keep Messages Clear and Consistent

Write logs that explain the issue, not just “Error occurred.”

Ex:logging.error("Database connection failed: timeout after 30s")

8.Use Module‑Level Loggers

Create a logger per module with __name__.

Ex:
python
logger = logging.getLogger(__name__)

9.Configure Logging Once

Use logging.config or a central setup

Avoid calling basicConfig() multiple times

10.Test Your Logging Setup

Run your app and check logs in both console and file

Ensure levels, formats, and rotation work as expected
```
