### Primary Key


```python
A Primary Key uniquely identifies each record in a table

Example:

CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50),
    age INT
);

student_id	name	age
1	        Vijay	20
2	        Vicky	21
3	        Ajay	20

We cannot have like these

1 Vijay
1 Vicky
because Student_id is a Primary Key
```

### Foreign Key

```python
A Foreign Key connects one table to another

Example:

CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50)
);

CREATE TABLE marks (
    mark_id INT PRIMARY KEY,
    student_id INT,
    marks INT,
    FOREIGN KEY (student_id)
        REFERENCES students(student_id)
);

Students

student_id	name
1       	Vijay
2	        Vicky

Marks

mark_id	student_id	marks
101       	1	    85
102	        2	    90

marks.student_id is the Foreign Key
It tells us that the marks belong to a particular student
```

### Constraints

```python
Constraints are rules applied to data

Common constraints:
    
PRIMARY KEY
FOREIGN KEY
NOT NULL
UNIQUE
CHECK
DEFAULT

Example:

CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE,
    age INT CHECK (age >= 18),
    city VARCHAR(50) DEFAULT 'Hyderabad'
);
```

### SELECT

```python
SELECT is used to retrieve data from a table

SELECT * FROM students; #Displays the student table

SELECT name, marks
FROM students;  #To get Specific Columns

OUTPUT will look like this

name	marks
Vijay	85
Vicky	90
Ajay	75
Bunny	90
Sai 	80
```

### WHERE

```python
WHERE is used to filter records based on a condition

SELECT *
FROM students
WHERE marks > 80;

OUTPUT Will look like this

Output:

id	name	marks
1	Vijay	85
2	Vicky	90
4	Bunny	90
```

### ORDER BY

```python
ORDER BY is used to sort the result

SELECT *
FROM students
ORDER BY marks ASC;

OUTPUT will look like this:

name	marks
Ajay	75
Sai 	80
Vijay	85
Vicky	90
Bunny	90

SELECT *
FROM students
ORDER BY marks DESC;

name	marks
Vicky	90
Bunny	90
Vijay	85
Sai 	80
Ajay	75
```

### GROUP BY

```python
GROUP BY is used to group rows having the same value

it is used with aggregate functions such as:

COUNT()
SUM()
AVG()
MAX()
MIN()

Example:

SELECT city, COUNT(*) AS total_students
FROM students
GROUP BY city;

OUTPUT will look like this:

Output:

city	total_students
Hyderabad	2
Banglore	2
Chennai	    1
```

### HAVING

```python
HAVING is used to filter groups created by GROUP BY

Show only cities having more than 1 students:

SELECT city, COUNT(*) AS total_students
FROM students
GROUP BY city
HAVING COUNT(*) > 1;

city	total_students
Hyderabad	2
Banglore	2
```

### DISTINCT

```python
DISTINCT is used to remove duplicate values

Our table has:

Hyderabad
Banglore
Hyderabad
Chennai
Banglore

Query:

SELECT DISTINCT city
FROM students;

Output:

city

Hyderabad
Banglore
Chennai
```

### LIMIT

```python
LIMIT is used to restrict the number of rows returned

SELECT *
FROM students
LIMIT 3;

Output:

id	name	marks
1	Vijay	85
2	Vicky	90
3	Ajay	75
```

## Joins

```python
A JOIN is used to combine data from two or more tables using a related column

Example:

students #Table-1

student_id	name	course_id
1	        Vijay	101
2	        Vicky	102
3	        Ajay	103
4	        Bunny	104

courses
course_id	course_name
101	         Python
102	         SQL
103	         ML
105	         Data Science

Here,
Student Bunny has course_id = 104, but there is no course 104
Data Science has course_id = 105, but no student has 105
```

### INNER JOIN


```python
INNER JOIN returns only matching records from both tables

SELECT students.name, courses.course_name
FROM students
INNER JOIN courses
ON students.course_id = courses.course_id;

OUTPUT will look like this

name	course_name
Vijay	Python
Vicky	SQL
Ajay	ML

Bunny is not included because 104 doesn't exist in courses

Data Science is not included because 105 doesn't exist in students
```

### LEFT JOIN


```python
LEFT JOIN returns all records from the left table and matching records from the right table

SELECT students.name, courses.course_name
FROM students
LEFT JOIN courses
ON students.course_id = courses.course_id;

OUTPUT will look like this:

name	course_name
Vijay	Python
Vicky	SQL
Ajay	ML
Bunny	NULL

Bunny is included because students is the left table

There is no matching course for Bunny, so course_name becomes NULL
```

### RIGHT JOIN

```python
RIGHT JOIN returns all records from the right table and matching records from the left table

SELECT students.name, courses.course_name
FROM students
RIGHT JOIN courses
ON students.course_id = courses.course_id;

OUTPUT will look like :
    
name	course_name
Vijay	Python
Vicky	SQL
Ajay	ML
NULL	Data Science

Data Science is included because courses is the right table

There is no student with course_id = 105, so name becomes NULL
```

### FULL JOIN

```python
FULL JOIN returns all records from both tables

It includes:

Matching records
Unmatched records from the left
Unmatched records from the right

SELECT students.name, courses.course_name
FROM students
FULL JOIN courses
ON students.course_id = courses.course_id;

OUTPUT will look like this:

name	course_name
Vijay	Python
Vicky	SQL
Ajay     ML
Bunny	NULL
NULL	Data Science

MySQL does not directly support FULL JOIN. It can be simulated using LEFT JOIN + UNION + RIGHT JOIN
```

### CROSS JOIN

```python
CROSS JOIN creates every possible combination of rows from the two tables

For example:

Students

name
Vijay
Vicky

Courses

course_name
Python

SELECT students.name, courses.course_name
FROM students
CROSS JOIN courses;

OUTPUT will look like this:

name	course_name
Vijay	Python
Vijay	SQL
Vicky	Python
Vicky	SQL
2 students × 2 courses = 4 combinations
```

### Advanced SQL

```python
Advanced SQL refers to more powerful SQL techniques used to perform complex data analysis, 
manipulate data efficiently, and retrieve meaningful information from databases

Example Table:

    Employee Table

id	name	department	salary
1	Vijay	 IT	        60000
2	Ajay	 IT       	70000
3	Bunny	 HR      	50000
4	Sai 	 HR	        60000
5	Kiran	 IT	        7000
```

### UNION

```python
UNION is used to combine the results of two SELECT queries

SELECT name FROM employees
WHERE department = 'IT'

UNION

SELECT name FROM employees
WHERE department = 'HR';

OUTPUT will look like this:

name
Vijay
Ajay
Bunny
Sai
kiran

UNION removes duplicates
```

### CASE

```python
CASE is used to create conditional logic in SQL

SELECT name, salary,
    CASE
        WHEN salary >= 70000 THEN 'High'
        WHEN salary >= 60000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_level
FROM employees;

OUTPUT will look like this:

name	salary	salary_level
Vijay	60000	Medium
Ajay	70000	High
Bunny	50000	Low
Sai 	60000	Medium
kiran	70000	High
```

### Subqueries

```python
A subquery is a query inside another query

SELECT name, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);

The inner query:

SELECT AVG(salary)
FROM employees;

calculates the average salary

The outer query finds employees whose salary is above that average
```

### CTEs

```python
CTE means Common Table Expression

A CTE creates a temporary named result that you can use in the main query

Syntax:

WITH name AS (
    SELECT ...
)
SELECT ...
FROM name;

Example:

WITH high_salary AS (
    SELECT name, salary
    FROM employees
    WHERE salary >= 60000
)
SELECT *
FROM high_salary;

OUTPUT look like this:

name	salary
Vijay	60000
Ajay	70000
Sai 	60000
Kiran	70000

Why use CTEs?

They make complex queries easier to read and understand
```

### Views

```python
A View is a saved SQL query that behaves like a virtual table

Creating a view:

CREATE VIEW it_employees AS
SELECT name, salary
FROM employees
WHERE department = 'IT';

SELECT *
FROM it_employees;

OUTPUT Will look like this:

name	salary
Vijay	60000
Ajay	70000
Kiran	70000
```

### Indexes

```python
An index helps the database find data faster

SELECT *
FROM employees
WHERE department = 'IT';

Creating an index:

CREATE INDEX idx_department
ON employees(department);

Now the database has an index that can help with searches involving department
```

## Window Functions

```python
Window functions perform calculations across related rows without combining those rows into a single row

This is an important difference from GROUP BY

Example:

SELECT name, department, salary,
       RANK() OVER (
           PARTITION BY department
           ORDER BY salary DESC
       ) AS salary_rank
FROM employees;
```

### ROW_NUMBER()

```python
ROW_NUMBER() gives each row a unique sequential number

SELECT name, department, salary,
       ROW_NUMBER() OVER (
           ORDER BY salary DESC
       ) AS row_num
FROM employees;

OUTPUT look like this:

name	salary	row_num
Ajay	70000	1
kiran	70000	2
Vijay	60000	3
Sain	60000	4
Bunny	50000	5
```

### RANK()

```python
RANK() gives the same rank to tied values, but skips numbers after a tie

SELECT name, salary,
       RANK() OVER (
           ORDER BY salary DESC
       ) AS salary_rank
FROM employees;

OUTPUT look like this:

name	salary	rank
Ajay	70000	1
Kiran	70000	1
Vijay	60000	3
Sai 	60000	3
Bunny	50000	5

RANK() = Same rank for ties + gaps
```

### DENSE_RANK()

```python
DENSE_RANK() also gives the same rank to tied values, but does not skip numbers

SELECT name, salary,
       DENSE_RANK() OVER (
           ORDER BY salary DESC
       ) AS salary_rank
FROM employees;

OUTPUT look like this:
    
name	salary	rank
Ajay	70000	1
kiran	70000	1
Vijay	60000	2
Sai 	60000	2
Bunny	50000	3

DENSE_RANK() = Same rank for ties + no gaps
```

### LAG()

```python
LAG() gets a value from a previous row

SELECT name, salary,
       LAG(salary) OVER (
           ORDER BY id
       ) AS previous_salary
FROM employees;

OUTPUT look like this:

name	salary	previous_salary
Vijay	60000	NULL
Ajay	70000	60000
Bunny	50000	70000
Sai 	60000	50000
Kiran	70000	60000

For Ajay:

Current salary = 70000
Previous salary = 60000

LAG() = Look backward
```

### LEAD()

```python
LEAD() gets a value from a next row

SELECT name, salary,
       LEAD(salary) OVER (
           ORDER BY id
       ) AS next_salary
FROM employees;

OUTPUT looks like this:

name	salary	next_salary
Vijay	60000	70000
Ajay	70000	50000
Bunny	50000	60000
Sai 	60000	70000
Kiran	70000	NULL

LEAD() = Look forward
```
