# Notebook 5 – File Handling

### Reading Text Files


```python
Methods to Read:
    
read() → Reads the entire file as one string

readline() → Reads one line at a time

readlines() → Reads all lines into a list
```

```python
#Example for Read entire file

with open("sample.txt", "w") as f:
    f.write("Hello Vijay!\nWelcome to File Handling.\nThis is line 3.")

with open("sample.txt", "r") as f:
    content = f.read()
    print("File Content:\n", content)
```

```python
#Example for Read one line at a time

with open("sample.txt", "r") as f:
    for line in f:
        print("Line:", line.strip())
```

```python
#example for Read all lines into a list

with open("sample.txt", "r") as f:
    lines = f.readlines()
    print("List of lines:", lines)


```

### Writing Text Files

```python
Writing a text file means creating a file or adding text into an existing file
```

```python
with open("hello.txt", "w") as file:
    file.write("Hello, Python!")
```

```python
Here:

open() → opens the file
"hello.txt" → file name
"w" → write mode
write() → writes text into the file
with → automatically closes the file
```

```python
with open("students.txt", "w") as file:          #Write multiple lines   #\n means new line
    file.write("Ajay\n")
    file.write("Vicky\n")
    file.write("Vijay\n")
```

### CSV Files


```python
CSV means Comma-Separated Values

A CSV file is used to store data in rows and columns, like a simple Excel table

Ex:
name,age,city
Ravi,25,Hyderabad
Vicky,27,Delhi
Kiran,24,Chennai
```

```python
import csv                          #Reading the CSV file

with open("students.csv", "r") as file:
    data = csv.reader(file)

    for row in data:
        print(row)
```

```python
import csv                        #Writing a CSV file

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "age"])
    writer.writerow(["Ravi", 25])
    writer.writerow(["Vicky", 27])
```

### JSON Files

```python
JSON means JavaScript Object Notation

JSON is commonly used to store and exchange data

json.dump() → Python data → JSON file
json.load() → JSON file → Python data
```

```python
import json                             #Create Jason data

employee = {
    "name": "Ravi",
    "age": 25,
    "job": "Python Developer",
    "salary": 30000
}

with open("employee.json", "w") as file:
    json.dump(employee, file)
```

```python
{                                            #This creates employee.json
    "name": "Ravi",                                
    "age": 25,
    "job": "Python Developer",
    "salary": 30000
}
```

```python
import json                          #Read the Json file

with open("employee.json", "r") as file:
    data = json.load(file)

print(data["name"])
print(data["job"])
```

### Pickle Files


```python
Pickle is used to save Python objects into a file and later load them back

It can save objects like lists, dictionaries, tuples, and custom objects
```

```python
#Example                #Creates data.pkl file

import pickle

data = ["Ravi", "Vicky", "Kiran"]

with open("data.pkl", "wb") as file:
    pickle.dump(data, file)
```

```python
import pickle                     #Read the pkl file

with open("data.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

```python
Explanation:-

pickle.dump() → save Python object
pickle.load() → load Python object
"wb" → write binary
"rb" → read binary
File extension commonly used: .pkl
```

### Context Manager (with)


```python
A context manager is used to manage resources automatically, such as files

The with statement automatically opens and closes the resource
```

```python
with open("test.txt", "w") as file:
    file.write("Hello World")
```

```python
with open("test.txt", "r") as file:
    data = file.read()
    print(data)
```

```python
with → manages the file
open() → opens the file
write() → writes text
File is automatically closed after the with block
with = automatically open → use → close
```

### pathlib

```python
pathlib is a Python module used to work with file and folder paths easily

Useful pathlib methods

Method  	Purpose

Path()  	Create a path
.exists()	Check whether path exists
.is_file()	Check if it is a file
.is_dir()	Check if it is a folder
.write_text()	Write text
.read_text()	Read text
.mkdir()	Create a folder
```

```python
from pathlib import Path           #Example

file = Path("hello.txt")

print(file.exists())
```

```python
from pathlib import Path             #Create a file

file = Path("hello.txt")

file.write_text("Hello Python")
```

```python
from pathlib import Path                 #Read a file

file = Path("hello.txt")

text = file.read_text()

print(text)
```

### shutil

```python
shutil is a built-in Python module used for copying, moving, and deleting files and folders

import shutil

shutil.copytree("photos", "photos_backup")
```

```python
What happens?

Before:

photos/
   image1.jpg
   image2.jpg

After:

photos/
   image1.jpg
   image2.jpg

photos_backup/
   image1.jpg
   image2.jpg

 shutil.copytree() copies an entire folder and its contents
```

### File Validation


```python
File validation means checking whether a file exists and whether it is valid for our requirements
```

```python
from pathlib import Path                #Example

file = Path("data.csv")

if file.exists() and file.is_file():
    print("File is valid")
else:
    print("File is not valid")
```

```python
from pathlib import Path

file = Path("data.csv")

if file.suffix == ".csv":
    print("Valid CSV file")
else:
    print("Invalid file type")
```

### Directory Operations

```python
Directory operations means creating, checking, listing, and deleting folders using Python

Python provides modules like os and pathlib for this
```

```python
import os

if os.path.exists("students"):
    print("Folder exists")
```

```python
import os

print(os.listdir())
```

```python
from pathlib import Path

folder = Path("students")

folder.mkdir(exist_ok=True)

print(folder.exists())
```
