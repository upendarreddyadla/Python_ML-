### Class


```python
A class is like a blueprint               #Understand the Concept
It defines what properties (variables) and behaviors (methods) an object will have, but it’s not the actual object itself
```

```python
Analogy: Think of a car design. The design describes the color, speed, and features, but it’s not a real car

The design = class, the real car = object          # Demonstrate the Concept
```

```python
class Student:                                                
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
class Student: → defines a blueprint for students                           #Explanation of a code

__init__ → constructor, runs automatically when you create an object

self.name and self.age → instance variables, unique for each student
```

### Object

```python
An object is a real instance created from a class         #Undestand the Concept
```

```python
Analogy: A red car built from the car design is an object    #Demonstrate the Concept
```

```python
s1 = Student("Ajay", 25)                 #S1 is Object name
print(s1.name, s1.age)
```

### Constructor (__init__)

```python
The constructor (__init__) runs automatically when an object is created. It initializes variables  #Understand the Concept
```

```python
Analogy: When you buy a phone, it comes pre‑set with default settings           #Demonstrate the Concept
```

```python
class Student:                                          #Implement the Concept
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Upendar", 30)
print(s1.name, s1.age)
```

```python
class Student:-This line defines a class called Student

__init__ is a special method in Python called the constructor It runs automatically whenever you create a new object of the class

Parameters:

self → refers to the current object being created.

name, age → values you pass when creating the object

Inside the constructor:

self.name = name → assigns the passed name to the object’s instance variable.

self.age = age → assigns the passed age to the object’s instance variable

print(s1.name, s1.age)-Accessing Instance Variables

def is a keyword

S1 is Object name
```

### Instance Variables

```python
Instance variables are variables that belong to a specific object

They are defined inside the constructor (__init__) using self

Each object has its own copy of these variables, so changing one object’s data does not affect another      #Understand the Concept
```

```python
Analogy: Imagine a school.

Each student has their own name and roll number.

These details are unique to each student.

That’s exactly how instance variables work — they store data separately for each object     #Demonstrate the Concept
```

```python
s1 = Student("Vijay", 25)                           #Implement the Concept
s2 = Student("Upendar", 30)

print(s1.name, s1.age)
print(s2.name, s2.age)
```

```python
s1 is an object with name = "Vijay" and age = 25

s2 is another object with name = "Upendar" and age = 30

s1.name → accesses Vijay’s name

s1.age → accesses Vijay’s age

s2.name → accesses Upendar’s name

s2.age → accesses Upendar’s age
```

### Class Variables

```python
Class variables are variables that are shared across all objects of a class    #understand the concept

They are defined inside the class but outside any methods

If one object changes the class variable, the change is reflected in all other objects because they all share the same copy   
```

```python
Analogy:-Imagine a class called Student                                #Demonstrate the Concept

Each student has their own name and age (instance variables)

But all students belong to the same school (class variable)

If the school name changes, it changes for all students at once
```

```python
class Student:

    school_name = "Hyderabad Public School"

    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    def show_details(self):
        print(f"Name: {self.name}, Age: {self.age}, School: {Student.school_name}")


s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)

s1.show_details()
s2.show_details()
Student.school_name = "Nalgonda School"

s1.show_details()
s2.show_details()
```

```python
school_name = "Hyderabad Public School"

This is a class variable.

It is defined inside the class but outside any methods.

All objects of the class share this variable.
    
self.name and self.age are instance variables → unique for each student.

s1 → Vijay, 25 years old.

s2 → Upendar, 30 years old.

Both share the same school_name.

1.show_details()
s2.show_details()
Both students show "ABC International School" because the class variable is shared

Student.school_name = "Nalgonda school"
We change the class variable at the class level.

Now, all objects automatically reflect the new school name.
```

### Instance Methods

```python
Instance methods are functions defined inside a class that operate on instance variables.   #Understand the Concept

They always take self as the first parameter, which refers to the specific object calling the method.

They allow each object to perform actions using its own data.
```

```python
Imagine a student in a school                                       #Demonstrate the Concept

Each student can “introduce themselves.”

When Vijay introduces himself, he uses his own name and age

When Upendar introduces himself, he uses his own name and age

The method is the same, but the data is different for each object
```

```python
class Student:                                            #Implement the Concept
    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    def show_details(self):   
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)

s1.show_details()
s2.show_details()
```

```python
show_details is an instance method because it uses self to access the object’s data            #Explanation

When s1.show_details() is called, self refers to s1, so it prints Vijay’s details

When s2.show_details() is called, self refers to s2, so it prints Upendar’s details

The same method works differently depending on which object calls it
```

### Static Methods


```python
A static method is a method inside a class that does not depend on instance variables (self) or class variables (cls)

It behaves like a normal function but is placed inside a class for better organization

You call it using the class name, and it doesn’t need an object to run
```

```python
Imagine a school                                         #Demonstrate the Concept

Each student has their own details (instance variables)

But sometimes, you just need a utility function like calculating the average of two marks

This calculation doesn’t depend on any particular student’s data

That’s what a static method does — a helper function inside the class
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):   
        return a + b

    @staticmethod
    def multiply(a, b):   
        return a * b

print(MathUtils.add(10, 20))
print(MathUtils.multiply(5, 4))
```

```python
class MathUtils: → defines a class named MathUtils (our utility class)              #Explanation

@staticmethod → tells Python that the method doesn’t use self (object data) or cls (class data)

add(a, b) → adds two numbers

multiply(a, b) → multiplies two numbers

We call them using MathUtils.add(10, 20) and MathUtils.multiply(5, 4) without creating an object
```

### Class Methods

```python
A class method is a method that works with class variables instead of instance variables     #Undestand the Concept

It takes cls as the first parameter (instead of self)

cls refers to the class itself, not a particular object

Class methods are useful when you want to perform actions that affect the whole class, not just one object
```

```python
Imagine a Student class                                              #Demonstrate the class

Each student has their own name and age (instance variables)

But all students share the same school name (class variable)

A class method can be used to access or change the school name for everyone at once
```

```python
class Student:                                                       #implementation
    school_name = "Hyderabad Public School"   

    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    @classmethod
    def get_school(cls):      
        return cls.school_name

    @classmethod
    def change_school(cls, new_name):   
        cls.school_name = new_name


s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)


print(s1.get_school())
print(s2.get_school())

Student.change_school("Nalgonda School")

print(s1.get_school())
print(s2.get_school())
```

```python
school_name → class variable, shared by all students                      #Explanation

@classmethod → decorator that makes the method a class method

get_school(cls) → returns the current school name

change_school(cls, new_name) → updates the school name for all students

Initially, both students show "ABC International School"

After calling Student.change_school("XYZ Global School"), both students reflect the new school name
```

### Properties (@property)

```python
A property in Python allows you to access methods like attributes                 #understand the concept

It’s mainly used for controlled access to private variables

Instead of calling a method like obj.get_marks(), you can just write obj.marks

This makes the code cleaner and safer, while still allowing validation or restrictions
```

Imagine a student’s marks                                      #Demonstrate the Concept

You don’t want anyone to directly change the marks variable

Instead, you provide a safe way to read the marks using a property

Later, you can also add rules (like marks must be between 0 and 100)

```python
class Student:                                                #implement the Concept
    def __init__(self, name, marks):
        self.name = name
        self._marks = marks   

    @property
    def marks(self):         
        return self._marks

    @marks.setter
    def marks(self, value):  
        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid marks! Must be between 0 and 100.")


s1 = Student("Vijay", 85)

print(s1.name, s1.marks)

s1.marks = 95
print(s1.name, s1.marks)

s1.marks = 120
```

```python
self._marks → private variable (not meant to be accessed directly)                 #Explanation

@property → allows you to access marks like an attribute (s1.marks)

@marks.setter → allows controlled updates. It checks if the new value is between 0 and 100

Initially, Vijay’s marks are 85

After updating, marks become 95

When trying to set 120, the setter rejects it with a warning
```

### Encapsulation 

```python
Encapsulation means hiding the internal details of how data is stored and controlled inside a class    #Undestand the concept

Instead of letting anyone directly change variables, we provide controlled access through methods or properties

This ensures data safety and prevents misuse
```

```python
Imagine a Bank Account                                                   #Demonstrate the Concept

You cannot directly open the bank’s database and change your balance

Instead, you use safe operations like deposit or withdraw

This ensures rules are followed (e.g., you can’t withdraw more than your balance)

That’s exactly how encapsulation works
```

```python
class BankAccount:                                               #Implement the Concept
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance   

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit must be positive!")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Insufficient funds!")

    def get_balance(self):
        return self.__balance  


account = BankAccount("Vijay", 100)
account.deposit(50)
account.withdraw(70)
print("Final Balance:", account.get_balance())


print(account.__balance)   
```

```python
self.__balance → private variable (hidden from direct access)
                                                                                     #Explanation 
You can’t directly change it (account.__balance gives an error)

You must use methods (deposit, withdraw, get_balance) to interact with the balance

This ensures the balance can only be modified in safe, controlled ways
```

### Abstraction

```python
Abstraction means hiding complex implementation details and showing only the essential features   #understand the concept

It lets the user interact with a simple interface without worrying about how things work internally

In Python, abstraction is often achieved using abstract classes and abstract methods (via the abc module)
```

```python

#Demonstrate the Concept

Imagine a Vehicle class

Every vehicle must have a way to start and stop, but the actual implementation differs (bike vs car)

Abstraction allows us to define the essential methods (start, stop) without worrying about the details

Each subclass (Car, Bike) provides its own implementation
```

```python
from abc import ABC, abstractmethod                #Implement the Concept


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car starts with a key ignition.")

    def stop(self):
        print("Car stops when brakes are applied.")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a kick or button.")

    def stop(self):
        print("Bike stops when brakes are pressed.")


v1 = Car()
v1.start()
v1.stop()

v2 = Bike()
v2.start()
v2.stop()
```

```python
#Explanation

class Vehicle(ABC) → abstract class (cannot be directly instantiated).

@abstractmethod → defines methods that must be implemented by subclasses.

Car and Bike → concrete classes that provide their own implementation of start and stop.

When we create Car and Bike objects, they behave differently, but the interface (start, stop) is the same.
```

### Inheritance

```python
Inheritance allows one class (child/subclass) to reuse the properties and methods of another class (parent/superclass)

It helps in code reusability and makes programs easier to maintain             #Understand the concept

The child class can also add new features or override parent features
```

```python
Imagine a Employee class (parent)                          #Demonstrate the Concept

Every employee has a name and salary

Now, we create a Manager class (child) that inherits from Employee

A manager has all employee features plus extra features like team size
```

```python

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}")

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        
        super().__init__(name, salary)
        self.team_size = team_size

    def show_details(self):
        
        print(f"Name: {self.name}, Salary: {self.salary}, Team Size: {self.team_size}")


emp1 = Employee("Vijay", 50000)
emp1.show_details()

mgr1 = Manager("Upendar", 80000, 10)
mgr1.show_details()
```

```python
Employee → parent class with name and salary                           #Explanation

Manager(Employee) → child class that inherits from Employee

super().__init__(name, salary) → calls the parent constructor to reuse code

Manager adds a new variable team_size

show_details in Manager overrides the parent method to show extra info

emp1 (Employee) shows only name and salary

mgr1 (Manager) shows name, salary, and team size
```

### Multiple Inheritance

```python
Multiple Inheritance means a child class can inherit features from more than one parent class     #Understand the Concept

This allows the child class to combine functionalities of multiple classes

Python supports multiple inheritance directly
```

```python
Imagine two classes:                         #Demonstrate the concept

Teacher → has teaching skills.

Cricketer → has cricket skills.

Now, we create a Person class that inherits from both.

This person can teach and also play cricket
```

```python

class Teacher:
    def teach(self):
        print("I can teach students.")


class Cricketer:
    def play_cricket(self):
        print("I can play cricket.")


class Person(Teacher, Cricketer):
    def __init__(self, name):
        self.name = name

    def show_details(self):
        print(f"My name is {self.name}.")


p1 = Person("Vijay")
p1.show_details()
p1.teach()         
p1.play_cricket()   
```

```python
Teacher → parent class with teach() method            #Explanation

Cricketer → parent class with play_cricket() method

Person(Teacher, Cricketer) → child class inherits from both

p1 object can access methods from both parent classes

This shows how multiple inheritance allows combining features
```

### Multilevel Inheritance

```python
Multilevel Inheritance means a child class inherits from a parent class, #understand the concept
and then another class inherits from that child class

It forms a chain of inheritance (Grandparent → Parent → Child)

Each level adds its own features, and the last child class has access to all features from the chain
```

```python
Imagine three classes:                                   #Demonstrate the Concept

Person → has basic details like name

Employee → inherits from Person and adds salary

Manager → inherits from Employee and adds team size

This shows how features pass down through multiple levels
```

```python

class Person:                                           #implement the code
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Name: {self.name}")


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)   
        self.salary = salary

    def show_salary(self):
        print(f"Salary: {self.salary}")


class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)   # call Employee constructor
        self.team_size = team_size

    def show_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}, Team Size: {self.team_size}")


m1 = Manager("Vijay", 80000, 10)
m1.show_name()      
m1.show_salary()    
m1.show_details()   
```

```python
#Explanation

Person → grandparent class with name

Employee(Person) → parent class inherits name and adds salary

Manager(Employee) → child class inherits both name and salary, and adds team_size

m1 object can access methods from all three levels
```

### Hierarchical Inheritance

```python
Hierarchical Inheritance means multiple child classes inherit from the same parent class       #Understand the concept

Each child class gets the parent’s features but can also add its own

This is useful when you have a common base class and different specialized subclasses
```

```python
Imagine a Person class (parent)                                 #Demonstrate the Concept

From it, we create two child classes: Teacher and Student

Both inherit the common features (like name), but each adds its own specific features
```

```python
                                                  #implement the Concept
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Name: {self.name}")


class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)   # call Person constructor
        self.subject = subject

    def show_details(self):
        print(f"Teacher Name: {self.name}, Subject: {self.subject}")


class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)   
        self.grade = grade

    def show_details(self):
        print(f"Student Name: {self.name}, Grade: {self.grade}")


t1 = Teacher("Vijay", "Mathematics")
s1 = Student("Upendar", "10th Grade")

t1.show_name()      
t1.show_details()   

s1.show_name()      
s1.show_details()   
```

```python
#Explanation:

Person → parent class with name

Teacher(Person) → child class inherits name and adds subject

Student(Person) → child class inherits name and adds grade

Both Teacher and Student share the parent’s show_name() method but have their own show_details() methods
```

### Polymorphism


```python
Polymorphism means many forms                       #Undestand the concept

In OOP, it allows the same method name to behave differently depending on the object that calls it

This makes code flexible and reusable
```

```python
Imagine different animals                          #Demonstrate the Concept

All animals can “speak,” but each speaks in its own way

A dog barks, a cat meows, a cow moos

The method name is the same (speak()), but the behavior changes depending on the object
```

```python
class Animal:                               #implement the Concept
    def speak(self):
        print("This animal makes a sound.")

class Dog(Animal):
    def speak(self):
        print("Dog says: Woof Woof!")

class Cat(Animal):
    def speak(self):
        print("Cat says: Meow Meow!")

class Cow(Animal):
    def speak(self):
        print("Cow says: Moo Moo!")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.speak()
```

```python
#Explanation:

Animal → parent class with a generic speak() method

Dog, Cat, Cow → child classes override speak() with their own behavior

We store all objects in a list and call animal.speak() in a loop

Even though the method name is the same, the output changes depending on the object type
```

### Method Overriding

```python
Method Overriding happens when a child class provides its own implementation of a method that already exists in the parent class

The method name and parameters are the same, but the behavior is different          #Understand the Concept

This is a key part of polymorphism in OOP
```

```python
Parent class Employee has a method work()               #Demonstrate the Concept

Child class Manager overrides work() because a manager’s work is different from a regular employee

Both have the same method name, but the output differs
```

```python

class Employee:                          #implement thr concept
    def work(self):
        print("Employee works on assigned tasks.")


class Manager(Employee):
    def work(self):   
        print("Manager oversees team and assigns tasks.")

class Developer(Employee):
    def work(self):   
        print("Developer writes and tests code.")


e1 = Employee()
m1 = Manager()
d1 = Developer()

e1.work()   
m1.work()   
d1.work()  
```

```python
#Explanation:-

Employee → parent class with work() method

Manager and Developer → child classes override work() with their own behavior

When calling work() on each object, Python decides which version to run based on the object type

This shows method overriding in action: same method name, different behavior
```

### Method Overloading (Python Approach)


```python
In many languages (like Java, C++), method overloading means having multiple methods with the same name but different parameter lists

Python does not support true method overloading — if you define multiple methods with the same name, the last one will overwrite the previous ones

Instead, Python achieves a similar effect using:                             #Understand the Concept

Default arguments

Variable-length arguments (*args, **kwargs)
```

```python
Think of a calculator’s add() function         #Demonstrate the concept

Sometimes you add two numbers, sometimes three, sometimes a list of numbers

Instead of writing separate add() methods, Python lets you write one flexible method that handles all cases
```

```python
class Calculator:                              #implement the Concept

    def add(self, a=0, b=0, c=0):
        return a + b + c


    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result


calc = Calculator()


print("Add 2 numbers:", calc.add(10, 20))


print("Add 3 numbers:", calc.add(10, 20, 30))


print("Add with no arguments:", calc.add())


print("Multiply 2 numbers:", calc.multiply(5, 4))


print("Multiply 4 numbers:", calc.multiply(2, 3, 4, 5))
```

```python
#Explanation

add(self, a=0, b=0, c=0) → uses default arguments to allow calling with 0, 2, or 3 arguments

multiply(self, *args) → uses variable-length arguments to allow multiplying any number of values

Python doesn’t create multiple versions of add() or multiply(). Instead, one flexible method handles all cases

This is how Python simulates method overloading
```
