# Notebook 2-Object-Oriented Programming

### Class


```python
A class is like a blueprint or template for creating objects                #understand the concept

It defines attributes (variables) and methods (functions) that describe how the object should behave

Objects are the actual instances created from a class

Syntax:-

class ClassName:
    
    def __init__(self, attribute1, attribute2):
        self.attribute1 = attribute1
        self.attribute2 = attribute2

    
    def show_details(self):
        print(f"Attribute1: {self.attribute1}, Attribute2: {self.attribute2}")
```

```python
class Car:                                        #implementation of the concept
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(f"The {self.color} {self.brand} is driving.")

car1 = Car("Toyota", "Red")
car2 = Car("Honda", "Blue")

car1.drive()
car2.drive()
```

```python
class Car: → defines the blueprint                         #explanation of a code

__init__() → constructor, runs when you create a new object

self.brand and self.color → attributes of the object

drive() → method that defines behavior

car1 and car2 → objects created from the class, each with its own values
```

### Object

```python
An object is a real instance created from a class (the blueprint)         #understand the concept

It has its own data (attributes) and can perform actions (methods) defined in the class

Each object is independent, even if created from the same class        
```

```python
class Car:                                                   #implementation of the concept
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(f"The {self.color} {self.brand} is driving.")


car1 = Car("Toyota", "Red")   
car2 = Car("Honda", "Blue")   

car1.drive()
car2.drive()

```

```python
#Explanation:-

class Car: → defines the blueprint.

car1 = Car("Toyota", "Red") → creates an object with brand Toyota and color Red.

car2 = Car("Honda", "Blue") → creates another object with different values
```

### Constructor (__init__)

```python
A constructor is a special method in Python called __init__     #Undestand the concept

It runs automatically whenever you create a new object from a class

Its job is to initialize the object’s attributes (set starting values)

Syntax:

class ClassName:
    def __init__(self, parameter1, parameter2):
        self.parameter1 = parameter1
        self.parameter2 = parameter2
        

        __init__ → constructor method.

self → refers to the current object being created.

Parameters → values you pass when creating the object.
```

```python
class Student:
    def __init__(self, name, age):   
        self.name = name             
        self.age = age

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}")


s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)

s1.show()
s2.show()
```

```python
#Explanation:

__init__ runs automatically when s1 and s2 are created

It assigns the values (name, age) to the object’s instance variables

Each object has its own data, but they share the same structure defined by the class
```

### Instance Variables

```python
Instance variables are variables that belong to a specific object  #Understand the concept

They are defined inside the constructor (__init__) using self

Each object has its own copy of these variables, so changing one object’s data does not affect another   

#Demonstrate the concept

 Imagine a school.

Each student has their own name and roll number.

These details are unique to each student.

That’s exactly how instance variables work — they store data separately for each object     
```

```python
Analogy: Imagine a school.

Each student has their own name and roll number.

These details are unique to each student.

That’s exactly how instance variables work — they store data separately for each object     #Demonstrate the Concept
```

```python
class Student:
    def __init__(self, name, roll_no):
        self.name = name        
        self.roll_no = roll_no 

    def show_details(self):
        print(f"Name: {self.name}, Roll No: {self.roll_no}")


s1 = Student("Vijay", 101)
s2 = Student("Upendar", 102)

s1.show_details()
s2.show_details()


s1.roll_no = 201

print("\nAfter updating Vijay's roll number:")
s1.show_details()
s2.show_details()

```

```python
self.name and self.roll_no → instance variables, unique for each object

s1 has "Vijay", 101 while s2 has "Upendar", 102

When we update s1.roll_no to 201, only Vijay’s roll number changes

Upendar’s roll number remains 102
```

### Class Variables

```python
Class variables are variables that are shared across all objects of a class    #understand the concept

They are defined inside the class but outside any methods

If one object changes the class variable, the change is reflected in all other objects because they all share the same copy   
```

```python
Analogy:-Imagine a class called Student                                #Demonstrate the Concept

Each student has their own name and age (instance variables)

But all students belong to the same school (class variable)

If the school name changes, it changes for all students at once
```

```python
class Student:

    school_name = "Hyderabad Public School"

    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    def show_details(self):
        print(f"Name: {self.name}, Age: {self.age}, School: {Student.school_name}")


s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)

s1.show_details()
s2.show_details()
Student.school_name = "Nalgonda School"

s1.show_details()
s2.show_details()
```

```python
school_name = "Hyderabad Public School"

This is a class variable.

It is defined inside the class but outside any methods.

All objects of the class share this variable.
    
self.name and self.age are instance variables → unique for each student.

s1 → Vijay, 25 years old.

s2 → Upendar, 30 years old.

Both share the same school_name.

1.show_details()
s2.show_details()
Both students show "ABC International School" because the class variable is shared

Student.school_name = "Nalgonda school"
We change the class variable at the class level.

Now, all objects automatically reflect the new school name.
```

### Instance Methods

```python
Instance methods are functions defined inside a class that operate on instance variables.   #Understand the Concept

They always take self as the first parameter, which refers to the specific object calling the method.

They allow each object to perform actions using its own data.
```

```python
Imagine a student in a school                                       #Demonstrate the Concept

Each student can “introduce themselves.”

When Vijay introduces himself, he uses his own name and age

When Upendar introduces himself, he uses his own name and age

The method is the same, but the data is different for each object
```

```python
class Student:                                            #Implement the Concept
    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    def show_details(self):   
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)

s1.show_details()
s2.show_details()
```

```python
show_details is an instance method because it uses self to access the object’s data            #Explanation

When s1.show_details() is called, self refers to s1, so it prints Vijay’s details

When s2.show_details() is called, self refers to s2, so it prints Upendar’s details

The same method works differently depending on which object calls it
```

### Static Methods


```python
A static method is a method inside a class that does not depend on instance variables (self) or class variables (cls)

It behaves like a normal function but is placed inside a class for better organization

You call it using the class name, and it doesn’t need an object to run
```

```python
Imagine a school                                         #Demonstrate the Concept

Each student has their own details (instance variables)

But sometimes, you just need a utility function like calculating the average of two marks

This calculation doesn’t depend on any particular student’s data

That’s what a static method does — a helper function inside the class
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):   
        return a + b

    @staticmethod
    def multiply(a, b):   
        return a * b

print(MathUtils.add(10, 20))
print(MathUtils.multiply(5, 4))
```

```python
class MathUtils: → defines a class named MathUtils (our utility class)              #Explanation

@staticmethod → tells Python that the method doesn’t use self (object data) or cls (class data)

add(a, b) → adds two numbers

multiply(a, b) → multiplies two numbers

We call them using MathUtils.add(10, 20) and MathUtils.multiply(5, 4) without creating an object
```

### Class Methods

```python
A class method is a method that works with class variables instead of instance variables     #Undestand the Concept

It takes cls as the first parameter (instead of self)

cls refers to the class itself, not a particular object

Class methods are useful when you want to perform actions that affect the whole class, not just one object
```

```python
Imagine a Student class                                              #Demonstrate the class

Each student has their own name and age (instance variables)

But all students share the same school name (class variable)

A class method can be used to access or change the school name for everyone at once
```

```python
class Student:                                                       #implementation
    school_name = "Hyderabad Public School"   

    def __init__(self, name, age):
        self.name = name      
        self.age = age        

    @classmethod
    def get_school(cls):      
        return cls.school_name

    @classmethod
    def change_school(cls, new_name):   
        cls.school_name = new_name


s1 = Student("Vijay", 25)
s2 = Student("Upendar", 30)


print(s1.get_school())
print(s2.get_school())

Student.change_school("Nalgonda School")

print(s1.get_school())
print(s2.get_school())
```

```python
school_name → class variable, shared by all students                      #Explanation

@classmethod → decorator that makes the method a class method

get_school(cls) → returns the current school name

change_school(cls, new_name) → updates the school name for all students

Initially, both students show "ABC International School"

After calling Student.change_school("XYZ Global School"), both students reflect the new school name
```

### Properties (@property)

```python
A property in Python allows you to access methods like attributes                 #understand the concept

It’s mainly used for controlled access to private variables

Instead of calling a method like obj.get_marks(), you can just write obj.marks

This makes the code cleaner and safer, while still allowing validation or restrictions
```

Imagine a student’s marks                                      #Demonstrate the Concept

You don’t want anyone to directly change the marks variable

Instead, you provide a safe way to read the marks using a property

Later, you can also add rules (like marks must be between 0 and 100)

```python
class Student:                                                #implement the Concept
    def __init__(self, name, marks):
        self.name = name
        self._marks = marks   

    @property
    def marks(self):         
        return self._marks

    @marks.setter
    def marks(self, value):  
        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid marks! Must be between 0 and 100.")


s1 = Student("Vijay", 85)

print(s1.name, s1.marks)

s1.marks = 95
print(s1.name, s1.marks)

s1.marks = 120
```

```python
self._marks → private variable (not meant to be accessed directly)                 #Explanation

@property → allows you to access marks like an attribute (s1.marks)

@marks.setter → allows controlled updates. It checks if the new value is between 0 and 100

Initially, Vijay’s marks are 85

After updating, marks become 95

When trying to set 120, the setter rejects it with a warning
```

### Encapsulation 

```python
Encapsulation means hiding the internal details of how data is stored and controlled inside a class    #Undestand the concept

Instead of letting anyone directly change variables, we provide controlled access through methods or properties

This ensures data safety and prevents misuse
```

```python
Imagine a Bank Account                                                   #Demonstrate the Concept

You cannot directly open the bank’s database and change your balance

Instead, you use safe operations like deposit or withdraw

This ensures rules are followed (e.g., you can’t withdraw more than your balance)

That’s exactly how encapsulation works
```

```python
class BankAccount:                                               #Implement the Concept
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance   

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit must be positive!")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Insufficient funds!")

    def get_balance(self):
        return self.__balance  


account = BankAccount("Vijay", 100)
account.deposit(50)
account.withdraw(70)
print("Final Balance:", account.get_balance())


print(account.__balance)   
```

```python
self.__balance → private variable (hidden from direct access)
                                                                                     #Explanation 
You can’t directly change it (account.__balance gives an error)

You must use methods (deposit, withdraw, get_balance) to interact with the balance

This ensures the balance can only be modified in safe, controlled ways
```

### Abstraction

```python
Abstraction means hiding complex implementation details and showing only the essential features   #understand the concept

It lets the user interact with a simple interface without worrying about how things work internally

In Python, abstraction is often achieved using abstract classes and abstract methods (via the abc module)
```

```python

#Demonstrate the Concept

Imagine a Vehicle class

Every vehicle must have a way to start and stop, but the actual implementation differs (bike vs car)

Abstraction allows us to define the essential methods (start, stop) without worrying about the details

Each subclass (Car, Bike) provides its own implementation
```

```python
from abc import ABC, abstractmethod                #Implement the Concept


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car starts with a key ignition.")

    def stop(self):
        print("Car stops when brakes are applied.")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a kick or button.")

    def stop(self):
        print("Bike stops when brakes are pressed.")


v1 = Car()
v1.start()
v1.stop()

v2 = Bike()
v2.start()
v2.stop()
```

```python
#Explanation

class Vehicle(ABC) → abstract class (cannot be directly instantiated).

@abstractmethod → defines methods that must be implemented by subclasses.

Car and Bike → concrete classes that provide their own implementation of start and stop.

When we create Car and Bike objects, they behave differently, but the interface (start, stop) is the same.
```

### Inheritance

```python
Inheritance allows one class (child/subclass) to reuse the properties and methods of another class (parent/superclass)

It helps in code reusability and makes programs easier to maintain             #Understand the concept

The child class can also add new features or override parent features
```

```python
Imagine a Employee class (parent)                          #Demonstrate the Concept

Every employee has a name and salary

Now, we create a Manager class (child) that inherits from Employee

A manager has all employee features plus extra features like team size
```

```python

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}")

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        
        super().__init__(name, salary)
        self.team_size = team_size

    def show_details(self):
        
        print(f"Name: {self.name}, Salary: {self.salary}, Team Size: {self.team_size}")


emp1 = Employee("Vijay", 50000)
emp1.show_details()

mgr1 = Manager("Upendar", 80000, 10)
mgr1.show_details()
```

```python
Employee → parent class with name and salary                           #Explanation

Manager(Employee) → child class that inherits from Employee

super().__init__(name, salary) → calls the parent constructor to reuse code

Manager adds a new variable team_size

show_details in Manager overrides the parent method to show extra info

emp1 (Employee) shows only name and salary

mgr1 (Manager) shows name, salary, and team size
```

### Multiple Inheritance

```python
Multiple Inheritance means a child class can inherit features from more than one parent class     #Understand the Concept

This allows the child class to combine functionalities of multiple classes

Python supports multiple inheritance directly
```

```python
Imagine two classes:                         #Demonstrate the concept

Teacher → has teaching skills.

Cricketer → has cricket skills.

Now, we create a Person class that inherits from both.

This person can teach and also play cricket
```

```python

class Teacher:
    def teach(self):
        print("I can teach students.")


class Cricketer:
    def play_cricket(self):
        print("I can play cricket.")


class Person(Teacher, Cricketer):
    def __init__(self, name):
        self.name = name

    def show_details(self):
        print(f"My name is {self.name}.")


p1 = Person("Vijay")
p1.show_details()
p1.teach()         
p1.play_cricket()   
```

```python
Teacher → parent class with teach() method            #Explanation

Cricketer → parent class with play_cricket() method

Person(Teacher, Cricketer) → child class inherits from both

p1 object can access methods from both parent classes

This shows how multiple inheritance allows combining features
```

### Multilevel Inheritance

```python
Multilevel Inheritance means a child class inherits from a parent class, #understand the concept
and then another class inherits from that child class

It forms a chain of inheritance (Grandparent → Parent → Child)

Each level adds its own features, and the last child class has access to all features from the chain
```

```python
Imagine three classes:                                   #Demonstrate the Concept

Person → has basic details like name

Employee → inherits from Person and adds salary

Manager → inherits from Employee and adds team size

This shows how features pass down through multiple levels
```

```python

class Person:                                           #implement the code
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Name: {self.name}")


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)   
        self.salary = salary

    def show_salary(self):
        print(f"Salary: {self.salary}")


class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)   # call Employee constructor
        self.team_size = team_size

    def show_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}, Team Size: {self.team_size}")


m1 = Manager("Vijay", 80000, 10)
m1.show_name()      
m1.show_salary()    
m1.show_details()   
```

```python
#Explanation

Person → grandparent class with name

Employee(Person) → parent class inherits name and adds salary

Manager(Employee) → child class inherits both name and salary, and adds team_size

m1 object can access methods from all three levels
```

### Hierarchical Inheritance

```python
Hierarchical Inheritance means multiple child classes inherit from the same parent class       #Understand the concept

Each child class gets the parent’s features but can also add its own

This is useful when you have a common base class and different specialized subclasses
```

```python
Imagine a Person class (parent)                                 #Demonstrate the Concept

From it, we create two child classes: Teacher and Student

Both inherit the common features (like name), but each adds its own specific features
```

```python
                                                  #implement the Concept
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Name: {self.name}")


class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)   # call Person constructor
        self.subject = subject

    def show_details(self):
        print(f"Teacher Name: {self.name}, Subject: {self.subject}")


class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)   
        self.grade = grade

    def show_details(self):
        print(f"Student Name: {self.name}, Grade: {self.grade}")


t1 = Teacher("Vijay", "Mathematics")
s1 = Student("Upendar", "10th Grade")

t1.show_name()      
t1.show_details()   

s1.show_name()      
s1.show_details()   
```

```python
#Explanation:

Person → parent class with name

Teacher(Person) → child class inherits name and adds subject

Student(Person) → child class inherits name and adds grade

Both Teacher and Student share the parent’s show_name() method but have their own show_details() methods
```

### Polymorphism


```python
Polymorphism means many forms                       #Undestand the concept

In OOP, it allows the same method name to behave differently depending on the object that calls it

This makes code flexible and reusable
```

```python
Imagine different animals                          #Demonstrate the Concept

All animals can “speak,” but each speaks in its own way

A dog barks, a cat meows, a cow moos

The method name is the same (speak()), but the behavior changes depending on the object
```

```python
class Animal:                               #implement the Concept
    def speak(self):
        print("This animal makes a sound.")

class Dog(Animal):
    def speak(self):
        print("Dog says: Woof Woof!")

class Cat(Animal):
    def speak(self):
        print("Cat says: Meow Meow!")

class Cow(Animal):
    def speak(self):
        print("Cow says: Moo Moo!")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.speak()
```

```python
#Explanation:

Animal → parent class with a generic speak() method

Dog, Cat, Cow → child classes override speak() with their own behavior

We store all objects in a list and call animal.speak() in a loop

Even though the method name is the same, the output changes depending on the object type
```

### Method Overriding

```python
Method Overriding happens when a child class provides its own implementation of a method that already exists in the parent class

The method name and parameters are the same, but the behavior is different          #Understand the Concept

This is a key part of polymorphism in OOP
```

```python
Parent class Employee has a method work()               #Demonstrate the Concept

Child class Manager overrides work() because a manager’s work is different from a regular employee

Both have the same method name, but the output differs
```

```python

class Employee:                          #implement thr concept
    def work(self):
        print("Employee works on assigned tasks.")


class Manager(Employee):
    def work(self):   
        print("Manager oversees team and assigns tasks.")

class Developer(Employee):
    def work(self):   
        print("Developer writes and tests code.")


e1 = Employee()
m1 = Manager()
d1 = Developer()

e1.work()   
m1.work()   
d1.work()  
```

```python
#Explanation:-

Employee → parent class with work() method

Manager and Developer → child classes override work() with their own behavior

When calling work() on each object, Python decides which version to run based on the object type

This shows method overriding in action: same method name, different behavior
```

### Method Overloading (Python Approach)


```python
In many languages (like Java, C++), method overloading means having multiple methods with the same name but different parameter lists

Python does not support true method overloading — if you define multiple methods with the same name, the last one will overwrite the previous ones

Instead, Python achieves a similar effect using:                             #Understand the Concept

Default arguments

Variable-length arguments (*args, **kwargs)
```

```python
Think of a calculator’s add() function         #Demonstrate the concept

Sometimes you add two numbers, sometimes three, sometimes a list of numbers

Instead of writing separate add() methods, Python lets you write one flexible method that handles all cases
```

```python
class Calculator:                              #implement the Concept

    def add(self, a=0, b=0, c=0):
        return a + b + c


    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result


calc = Calculator()


print("Add 2 numbers:", calc.add(10, 20))


print("Add 3 numbers:", calc.add(10, 20, 30))


print("Add with no arguments:", calc.add())


print("Multiply 2 numbers:", calc.multiply(5, 4))


print("Multiply 4 numbers:", calc.multiply(2, 3, 4, 5))
```

```python
#Explanation

add(self, a=0, b=0, c=0) → uses default arguments to allow calling with 0, 2, or 3 arguments

multiply(self, *args) → uses variable-length arguments to allow multiplying any number of values

Python doesn’t create multiple versions of add() or multiply(). Instead, one flexible method handles all cases

This is how Python simulates method overloading
```

### Composition

```python
Composition means building a class using objects of other classes          #understand the concept

Instead of inheriting, you combine classes to form a “has‑a” relationship

```

```python
A Car cannot run without an Engine                                #Demonstrate the concept

Instead of making Car inherit from Engine (which doesn’t make sense), we say:

Car has an Engine

This is composition — one class contains another class as a part
```

```python
                                                           #implement the concept
class Engine:
    def start(self):
        print("Engine starts...")

    def stop(self):
        print("Engine stops...")


class Car:
    def __init__(self, brand):
        self.brand = brand
        self.engine = Engine()   

    def drive(self):
        self.engine.start()
        print(f"{self.brand} car is driving...")
        self.engine.stop()


c1 = Car("Toyota")
c1.drive()
```

```python
#Explanation:-

Engine → separate class with start() and stop() methods

Car → contains an Engine object inside (self.engine = Engine())

When drive() is called, the Car uses its Engine to start, drive, and stop

This shows composition: Car is built using Engine, but Car is not an Engine
```

### Aggregation

```python
Aggregation is a special form of association between classes                 #Understand the concept

It represents a “has‑a” relationship, but unlike composition, the child object can exist independently of the parent
```

```python
A Team class contains multiple Player objects              #Demonstrate the concept

Players can exist independently (they don’t depend on the team to exist)

The team just uses them
```

```python
                                                                                #Implement the concept
class Player:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(f"Player: {self.name}")


class Team:
    def __init__(self, team_name, players):
        self.team_name = team_name
        self.players = players   

    def show_team(self):
        print(f"Team: {self.team_name}")
        for player in self.players:
            player.show()


p1 = Player("Vijay")
p2 = Player("Upendar")
p3 = Player("Rahul")


team = Team("Sunrisers Hyderabad", [p1, p2, p3])
team.show_team()
```

```python
#Explanation:-

Player → independent class with name

Team → contains a list of Player objects

Players (p1, p2, p3) are created independently

The Team object uses these players but doesn’t own their lifecycle

Even if the Team object is deleted, the Player objects still exist
```

### Association

```python
Association is a general relationship between two classes            #Understand the concept

It simply means one class uses another class

Unlike inheritance, association does not imply “is‑a” but rather “works with” or “uses.”

Association can be of two types:

Aggregation (weak relationship, objects can exist independently)

Composition (strong relationship, objects depend on the parent)

So, association is the umbrella concept, and aggregation/composition are its special cases
```

```python
A Teacher is associated with a Department                #Demonstrate the concept

The teacher works in the department, but both can exist independently

This is a simple association
```

```python
                                                        #implement the concept
class Department:                 
    def __init__(self, dept_name):
        self.dept_name = dept_name

    def show_department(self):
        print(f"Department: {self.dept_name}")


class Teacher:
    def __init__(self, name, department):
        self.name = name
        self.department = department   # association

    def show_details(self):
        print(f"Teacher: {self.name}")
        self.department.show_department()


dept = Department("Computer Science")


t1 = Teacher("Vijay", dept)
t1.show_details()
```

```python
#Explanation:-

Department → independent class

Teacher → associated with Department via an object reference

The teacher uses the department object but does not own it

Both can exist independently, but they are linked through association
```

### Magic (Dunder) Methods

```python
Magic methods (also called dunder methods, because they start and end with double underscores __) are special methods in Python

They let you define how objects of your class behave with built‑in operations (like +, len(), str(), comparisons, etc.)

Example: __init__, __str__, __len__, __add__, etc

They make your custom classes act more like built‑in types           #understand the Concept
```

```python
Think of a Book class                                                     #Demonstrate the concept

Normally, if you print a book object, you’ll see something like <Book object at 0x...>

But if you define __str__, you can control what gets printed (like title and author)

Similarly, __len__ can define what len(book) should return (like number of pages)
```

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):   
        return f"'{self.title}' by {self.author}"

    def __len__(self): 
        return self.pages

    def __add__(self, other):   
        return self.pages + other.pages


b1 = Book("Python Basics", "Vijay", 300)
b2 = Book("Advanced Python", "Upendar", 400)

print(b1)            
print(len(b1))       
print(b1 + b2)       
```

```python
#Explanation:-

__init__ → constructor, initializes object attributes

__str__ → defines how the object is printed

__len__ → defines what len(object) returns

__add__ → defines behavior of + between two objects

This makes custom objects behave like built‑in Python types
```

### Super()

```python
super() is a built‑in function in Python used inside a child class to call methods from the parent class

It helps in reusing parent class code without rewriting it

Commonly used in inheritance when overriding methods                               #Understand the concept
```

```python
Imagine a Person class with basic details                           #Demonstrate the concept

An Employee class inherits from Person

Instead of rewriting the code to set the name, Employee can use super() to call the parent’s constructor and then add salary details
```

```python
                                                             #implement the concept
class Person:
    def __init__(self, name):
        self.name = name
        print(f"Person constructor called for {self.name}")

    def show(self):
        print(f"Name: {self.name}")


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)   
        self.salary = salary
        print(f"Employee constructor called for {self.name}")

    def show(self):
        super().show()   # call parent method
        print(f"Salary: {self.salary}")


e1 = Employee("Vijay", 50000)
e1.show()
```

#Explanation:-

super().__init__(name) → calls the parent (Person) constructor to set the name

super().show() → calls the parent’s show() method to print the name

The child class (Employee) adds its own logic (salary) while still reusing parent code

This avoids duplication and makes inheritance cleaner

### __str__()

```python
__str__() is a magic (dunder) method in Python                         #Understand the concept

It defines the string representation of an object when you use print() or str()

Without __str__(), printing an object shows something like <ClassName object at 0x...>

With __str__(), you can control what meaningful text is displayed
```

```python
Imagine a Book object                              #Demonstrate the concept

If you print it without __str__(), you’ll just see a technical ID

But with __str__(), you can make it show "Book: Python Basics by Vijay"

Much more readable and useful
```

```python
class Book:                                                   #implement the concept
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):   
        return f"Book: '{self.title}' by {self.author}"


b1 = Book("Python Basics", "Vijay")
print(b1)          # calls __str__()
print(str(b1))     # also calls __str__()
```

```python
#Explanation:-

__str__() → defines how the object is represented as a string

When you call print(b1) or str(b1), Python automatically calls b1.__str__()

Instead of showing <Book object at 0x...>, it shows a meaningful description

This is very useful for debugging and user‑friendly output
```

### __repr__()

```python
__repr__() is a magic (dunder) method in Python             #understand the concept

It defines the official string representation of an object, mainly used for debugging and developers

When you call repr(obj) or just type the object in the Python shell, __repr__() is invoked
```

```python
Imagine a Book object                                        #Demonstrate the concept

__str__() shows a user‑friendly description like "Book: Python Basics by Vijay"

__repr__() shows a developer‑friendly description like "Book('Python Basics', 'Vijay')" which looks like code to recreate the object
```

```python
class Book:                                                  #implement the concept
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):   
        return f"'{self.title}' by {self.author}"

    def __repr__(self):  
        return f"Book('{self.title}', '{self.author}', {self.pages})"


b1 = Book("Python Basics", "vijay", 300)

print(str(b1))   
print(repr(b1))  
```

```python
#Explanation:-
__str__() → returns a readable string for end users

__repr__() → returns a developer string that shows how the object could be recreated

print(str(b1)) → 'Python Basics' by Vijay (friendly)

print(repr(b1)) → Book('Python Basics', 'vijay', 300) (technical)
```

### __len__()


```python
__len__() is a magic (dunder) method in Python        #understand the concept

It defines what the built‑in len() function should return for your custom object

By default, len() works on lists, strings, tuples, etc

But with __len__(), you can make your own classes behave like those built‑in types
```

```python
Imagine a ShoppingCart class                         #Demonstrate the concept

Normally, Python doesn’t know how to calculate its length

But if you define __len__(), you can make len(cart) return the number of items inside
```

```python
class ShoppingCart:                                   #implement the code
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def __len__(self):  
        return len(self.items)


cart = ShoppingCart()
cart.add_item("Laptop")
cart.add_item("Mouse")
cart.add_item("Keyboard")

print("Items in cart:", len(cart)) 
```

```python
#Explanation:-

__len__() → tells Python how to calculate the length of the object

In this example, it returns the length of the items list

When we call len(cart), Python internally calls cart.__len__()

This makes the custom class behave like a list or string with respect to length
```

### Dataclasses (Bonus)

```python
A dataclass is a Python decorator (@dataclass) introduced in Python 3.7.       #understand the concept

It automatically generates common methods for classes like:

__init__() → constructor

__repr__() → developer-friendly representation

__eq__() → equality comparison

This reduces boilerplate code when you just want to store data
```

```python
Normally, if you create a Student class, you’d have to manually write the constructor, __repr__, and maybe __eq__

With @dataclass, Python does all that for you automatically

It’s like hiring an assistant to write repetitive code for you             #Demonstrate the class
```

```python
from dataclasses import dataclass          #implement the concept

@dataclass
class Student:
    name: str
    grade: int
    age: int


s1 = Student("Vijay", 10, 15)
s2 = Student("Upendar", 10, 15)

print(s1)            
print(s1 == s2)  
```

```python
#Explanation:-

@dataclass → decorator that tells Python to auto-generate methods

__init__() → automatically created to set name, grade, and age

__repr__() → automatically created to show a readable representation

__eq__() → automatically created to compare objects by their values

We didn’t write any of these methods manually — Python did it for us
```

### Singleton Pattern (Bonus)


```python
The Singleton Pattern ensures that a class has only one instance throughout the program     #undestand the concept

No matter how many times you try to create the object, you’ll always get the same instance

Useful when you need a single shared resource (like a database connection, configuration manager, or logging system)
```

```python
Imagine a DatabaseConnection class              #Demonstrate the concept

You don’t want multiple connections being created unnecessarily

The Singleton ensures only one connection object exists, and everyone uses it
```

```python
class Singleton:               #implementation of the concept
    _instance = None   

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:   
            cls._instance = super().__new__(cls)
        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print("Object 1:", obj1)
print("Object 2:", obj2)
print("Are both same?", obj1 is obj2)
```

```python
#Explanation:-

__new__() → controls object creation.

_instance → stores the single instance of the class.

If _instance is None, a new object is created. Otherwise, the existing one is returned.

Both obj1 and obj2 point to the same memory location.
```
