# Notebook-3 Exception Handling

Exception Handling is a way to manage errors in your program so it doesn’t crash

An exception is an error that occurs during program execution (like dividing by zero, invalid input, or missing files)

With exception handling, you can catch these errors and respond gracefully instead of letting the program stop


### Try

```python
The try block is used for exception handling                       #understand the Concept

It lets you test a block of code for errors

If an error occurs, Python jumps to the except block instead of crashing the program

You can also use:

finally → code that always runs (cleanup tasks)

else → code that runs if no error occurs

Syntax:

try:
    # code that may cause an error
except:
    # code to handle the error
else:
    # code if no error occurs
finally:
    # code that always runs
```

```python
#Example

try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid numbers!")

else:
    print("Division successful!")

finally:
    print("Program finished.")
```

```python
Code Explanation:

try → runs the risky code (division)

except ZeroDivisionError → catches division by zero

except ValueError → catches invalid input (like text instead of numbers)

else → runs only if no error occurs

finally → runs no matter what (cleanup, closing files, etc.)
```

### Except

```python
except is part of Python’s exception handling system                #understand the Concept

It works with the try block

If an error occurs inside try, Python jumps to the matching except block

Each except can handle a specific type of error (like ZeroDivisionError, ValueError, etc.)

Syntax:-

try:
    # risky code
except ErrorType:
    # code to handle that error
```

```python
#Example

try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:   
    print("Error: Cannot divide by zero!")

except ValueError:          
    print("Error: Please enter valid numbers!")
```

```python
Code Explanation:

try → runs the risky code

except ZeroDivisionError → handles division by zero

except ValueError → handles invalid input

If no error occurs, the program just runs normally
```

### Else

```python
The else block is used together with try and except

It runs only if no exception occurs inside the try block

Think of it as:

try → risky code

except → rescue team if error happens

else → celebration code if everything goes fine

Syntax:

try:
    # risky code
except SomeError:
    # handle error
else:
    # runs only if no error occurs
```

```python
#Example

try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid numbers!")

else:
    print("Division successful! No errors occurred.")
```

```python
code Explanation:

try → runs risky code (division)

except → catches errors (like divide by zero or invalid input)

else → runs only if no error happened in try
```

### finally

```python
The finally block is used with try and except                #understands the concept

It runs no matter what happens — whether an error occurs or not

Commonly used for cleanup tasks (like closing files, releasing resources, disconnecting databases)

 Syntax:

    try:
    # risky code
except SomeError:
    # handle error
finally:
    # always runs
              
```

```python
#Example 

try:
    f = open("data.txt", "r")
    content = f.read()
    print(content)

except FileNotFoundError:
    print("Error: File not found!")

finally:
    print("Closing file...")
    try:
        f.close()
    except:
        print("File was never opened.")
```

```python
Code Explanation:

try → attempts to open and read the file

except FileNotFoundError → handles missing file error

finally → runs cleanup code (closing the file), regardless of success or failure
```

### raise


```python
The raise keyword is used to manually throw an exception in Python      #understand the concept

Normally, exceptions happen automatically (like dividing by zero)

But sometimes you want to force an error if certain conditions are not met

This is useful for validation and custom error handling

Syntax:

    raise ExceptionType("Error message")
```

```python
#Example

def register_student(name, age):
    if age < 5:
        raise ValueError("Age must be at least 5 to register.")  
    print(f"Student {name} registered successfully!")

try:
    register_student("Vijay", 3)
except ValueError as e:
    print("Registration failed:", e)
```

```python
Code Explanation:

raise → used to trigger an exception intentionally

Can use built‑in exceptions (ValueError, TypeError, etc.)

Can define custom exceptions for special cases

Helps enforce rules and validations in your program
```

### Custom Exceptions

```python
Python has built‑in exceptions (ValueError, TypeError, ZeroDivisionError, etc.)

But sometimes you need to define your own exception type for special cases in your program

A custom exception is just a class that inherits from Python’s built‑in Exception class

This makes your code more readable and specific when handling errors.

    Why use?

Makes error messages clearer and specific.

Helps in large projects where different modules may need their own error types.

Improves debugging because you can catch your custom exception separately from built‑in ones.

Syntax:

class MyCustomError(Exception):
    pass
```

```python
#Example:


class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        super().__init__(f"Balance {balance} is not enough to withdraw {amount}")


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)  
    balance -= amount
    print(f"Withdrawal successful! Remaining balance: {balance}")
    return balance


try:
    withdraw(1000, 1500)
except InsufficientFundsError as e:
    print("Transaction failed:", e)
```
